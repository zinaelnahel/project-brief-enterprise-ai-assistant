# IronStore: document inspection walkthrough

The production app uses Next.js. This notebook inspects its source documents using the same PDF loader as the corpus exporter. It does not maintain a second assistant implementation. Run from the repository root after installing requirements.txt.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / "scripts"))
from pdf_documents import load_documents
chunks, inventory = load_documents()
print(f"{len(inventory)} PDFs; {len(chunks)} passages")
```

## Inspect the collection

Review departments, extraction quality, and explicitly historical documents before enabling retrieval.

```python
from collections import Counter
print(Counter(d.get("department") for d in inventory))
print("Historical:", [d["source"] for d in inventory if d.get("historical")])
print("Empty pages:", [d for d in inventory if d.get("empty_pages")])
```

## Inspect source provenance

Each passage preserves its filename, page, department, and stable ID. Read the text to check policy dates and authority.

```python
sample = next(c for c in chunks if "Expense_Reimbursement" in c.source)
print(sample.source, "Page", sample.page, "ID", sample.id)
print(sample.text)
```
